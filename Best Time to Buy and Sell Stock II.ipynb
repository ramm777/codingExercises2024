- When summing all positive differences, you effectively capture every single profitable opportunity, no matter how small. This approach allows you to take advantage of every upward trend in the stock prices, leading to a maximized cumulative gain.
- This sum will always be the maximum!

```python
prices = [7,1,0,0,2,40]
prices = [7,6,4,3,1]

total_profit = 0
diffs = [] 
for i in range(1, len(prices)):
    
    diff = prices[i] - prices[i - 1]
    diffs.append(diff) 
    if diff >= 0:
        total_profit = total_profit + diff

total_profit, diffs         
```
