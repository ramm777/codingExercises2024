- Strategy - As you iterate, you keep minimizing min_price, and maximizing max_profit, by finding them as you iterate   

```python
def maxProfit(prices):
    
    min_price  = prices[0] # just randomly initialize
    max_profit = 0

    # As you iterate, you keep minimizing min_price, and maximizing max_profit, by finding them as you iterate    
    for price in prices:
        
        if price < min_price: # Keep finding min as you iterate
            min_price = price
        
        profit = price - min_price
        if profit > max_profit:
            max_profit = profit
        print('price:', price, ' min_price',  min_price, 'max_profit', max_profit)    
    
    return max_profit

prices1 = [7, 1, 5, 3, 6, 4]

print(maxProfit(prices1))  # Output: 5
```
