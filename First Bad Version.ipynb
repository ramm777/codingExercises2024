AK problem explanation:
- Given the total number of versions possible "n" and isBadVersion(version) function that return True/False
- Find the first bad version 
- e.g. n = 5, versions possible [1,2,3,4,5]
- Your are not given, which exactly version is bad, only the function that provides you answer for query - isBadVersion(version)

```python
# Using binary search

bad = 4
n   = 50

def isBadVersion(version):
    # e.g. bad = 4 returns [F,F,F,T,T... 50T]
    return version >= bad

# versions [1,2,3,4,5,6,7..., 25, ...50]
#                              | try mid 
# if mid True, then after mid - all are true to the end (as they bigger) - hence can cut out 
# But it may not be the first bad
# Again calc mid and try, can be False for example. Then move "left" to find the first True 
# left will eventually NOT be left < right, that is where you return right (or left), as it's boundary first True

def firstBadVersion(n): 
    left  = 1
    right = n

    while left < right:
        mid = (left + right) // 2
        print("[", left, "...,...", right, "]")
        if isBadVersion(mid) is True:
            right = mid
        else:
            left = mid + 1
    
    return right

firstBadVersion(n)
```
