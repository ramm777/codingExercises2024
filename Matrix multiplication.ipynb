```python
# O(N^3)

def matrix_multiply(A, B):
    
    m = [[0 for _ in range(len(B[0]))] for _ in range(len(A))]
    
    for i in range(len(A)):  # 2
        for j in range(len(B[0])): # 4
            for k in range(len(B)):  # 3 # Accumulation (summation) happens here
                m[i][j] = m[i][j] + A[i][k] * B[k][j]
    
    return m
```

```python
import numba
from numba import prange
```

- numba can compile Python functions to optimized machine code
- optimized machine code refers to code that has been translated into the lowest level of instructions that a computer's processor can execute directly. This process involves converting high-level programming languages like Python into machine code that is highly optimized for performance on specific hardware
- prange to parallelize the loops. I have only parallelized the outer loop, but you can also apply this to the inner loops.

```python
%%time # 1.6s

@numba.njit(parallel=True)
def matrix_multiply_parallel(A, B):

    rows_A, cols_A = A.shape
    rows_B, cols_B = B.shape

    assert cols_A == rows_B, "Incompatible matrices"

    m = np.zeros((rows_A, cols_B), dtype=np.float64)

    for i in prange(rows_A):  # Parallelize over rows of A
        for j in range(cols_B):  # Iterate over columns of B
            for k in range(cols_A):  # Or rows_B
                m[i, j] += A[i, k] * B[k, j]

    return m

matrix_multiply_parallel(A, B)
```
