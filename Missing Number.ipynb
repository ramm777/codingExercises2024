# Exclusive Or - XOR
 


```python
# Just learn by heart, it's a difficult logic

def missingNumber(nums):
    
    xor_sum = 0
    
    # By XORing all numbers from 0 to n, we create a complete sequence of XOR operations that would represent a scenario 
    # where no number is missing
    for i in range(len(nums) + 1):
        xor_sum = xor_sum ^ i
        
    # When we XOR this sequence with all elements in the array, every number that appears in both sequences will cancel out
    # due to the property A ^ A = 0
    for num in nums:
        xor_sum = xor_sum ^ num

    # The number that does not have a pair in the array (i.e., the missing number) will not cancel out

    return xor_sum


nums = [9,6,4,2,3,5,7,0,1]
missingNumber(nums)  
```
