```python
class ListNode: 
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

# head = [7, 8, 9, 8, 7]
head = ListNode(7)
head.next = ListNode(8)
head.next.next = ListNode(9)
head.next.next.next = ListNode(8)
head.next.next.next.next = ListNode(7)
```

```python
queue = []
current = head
while current: 
    queue.append(current.val)
    current = current.next
queue    
```

```python
i = 0
j = len(queue)-1 
while i < j: 
    if queue[i] != queue[j]: 
        print(False)
    i+=1
    j-=1
print(True)            
```

```python
# O(n) space due to the additional list (queue).

def isPalindrome(head):
    queue = []
    current = head
    while current: 
        queue.append(current.val)
        current = current.next

    i = 0
    j = len(queue)-1 
    while i < j: 
        if queue[i] != queue[j]: 
            return False
        i+=1
        j-=1
    return True      

isPalindrome(head)
```
