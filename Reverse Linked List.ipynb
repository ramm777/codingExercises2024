```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)
head.next.next.next = ListNode(4)
head.next.next.next.next = ListNode(5)
```

```python
def print_list(node):
    while node is not None:
        print(node.val, end=" -> ")
        node = node.next
    print("None")
    
print_list(head)
```

```python
class Solution: 
    def reverseList(self, head):
        
        if not head:
            return None
        
        queue = []
        current = head
        while current:
            queue.append(current)
            current = current.next
        
        new_head = queue.pop(-1)
        current = new_head
        while queue:
            current.next = queue.pop(-1)
            current = current.next
        
        current.next = None
        
        return new_head

solution = Solution()
reversed_head = solution.reverseList(head)
print_list(reversed_head)
```
