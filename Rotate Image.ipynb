```python
m = [[1,2,3],[4,5,6],[7,8,9]]

def RotateImage(m): 
    
    if not m: 
        return []

    for i in range(len(m)): # down 
        for j in range(i, len(m[0])): # right from [i to len]
            
            m[i][j], m[j][i] = m[j][i], m[i][j]

    for i in range(len(m)):
        m[i] = m[i][::-1]       

    return m       

RotateImage(m)  
```
