```python
# This is a very elegant way - but memory limit is exceeded

def quicksort(arr):
    
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2]  # For simplicity I use Middle Element, not Median-of-three

    left   = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right  = [x for x in arr if x > pivot]
    
    return quicksort(left) + middle + quicksort(right)

arr = [3, 6, 8, 10, 1, 2, 1] 
display(quicksort(arr)) 

arr = [9,-3,5,2,6,8,-6,1,3]
display(quicksort(arr))
```
