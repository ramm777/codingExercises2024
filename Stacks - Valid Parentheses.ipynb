```python
#s = "[])"
#s = "([]{)}"
s = "([]{})"
#s = ")("

s = list(s)

# Notice, indexing corresponds left_side to right_side
left_side  = ["(", "{", "["] 
right_side = [")", "}", "]"]

def isValid(s):

    if not s: return False

    stack = []
    for i in s:
        if i in left_side: 
            stack.append(i)
        
        elif i in right_side: 
            pos  = right_side.index(i) 
            
            if len(stack)==0:                 # if it's only right side - False
                return False 
            elif left_side[pos] != stack[-1]: # If parethesis do not close {} - False 
                return False
            else:                             # If parenthesis close - remove last from stack    
                stack.pop(-1)
        print(stack)

    if len(stack) != 0:
        return False                

    return True

print(isValid(s))
```
