```python
nums = [-1,0,3,5,9,12]
target = 9

def binary_search(nums, target): 
    lo = 0
    hi = len(nums) - 1

    while lo <= hi: 
        mid = (lo + hi) // 2

        if nums[mid] == target: 
            return mid 
        elif nums[mid] < target: 
            lo = mid + 1
        else: 
            hi = mid - 1
    
    return -1 # fallback if not found                     

binary_search(nums, target)
```
