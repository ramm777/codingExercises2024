# This is a PwC challange




- Memorize the solution of a gradient of Loss wrt a and b:



```python
import math

def ArrayChallenge(arr):
    x, y, a, b = arr
    learning_rate = 1
    
    y_pred = 1 / (1 + math.exp(-a * x - b))
    
    gradient_a = - (y - y_pred) * x
    gradient_b = - (y - y_pred)
    
    a_new = a + learning_rate * gradient_a
    b_new = b + learning_rate * gradient_b
    
    return "%0.3f" %a_new, "%0.3f" %b_new

# Example usage
print(ArrayChallenge([1, 1, 1, 1]))         # Expected output: 0.881, 0.881
print(ArrayChallenge([2.2, 0.0, 5.1, 5.7])) # Expected output: 7.300, 6.700
```
