```python
def findPeakElement(nums):
    left, right = 0, len(nums) - 1
    
    while left < right:
        mid = (left + right) // 2
        print(f"left: {left}, right: {right}, mid: {mid}, nums[mid]: {nums[mid]}, nums[mid + 1]: {nums[mid + 1]}")
        
        if nums[mid] > nums[mid + 1]:
            # peak could be the current mid element itself or further to the left
            right = mid
            print(f"nums[mid] > nums[mid + 1], move right to {right}")
        else:
            # peak could be on the right side (excluding mid)
            left = mid + 1
            print(f"nums[mid] <= nums[mid + 1], move left to {left}")
    
    # When left == right, we have found the peak element
    print(f"Found peak at index {left}, value: {nums[left]}")
    return left

nums2 = [1, 2, 1, 3, 5, 6, 4]
findPeakElement(nums2)  
print('-----------------------')
nums2 = [1, 2, 1, 3, 5, 1, 4]  # This looks wrong, as you need two neighbours?! 
findPeakElement(nums2)  
```

The binary search algorithm works efficiently by narrowing down the search range based on local comparisons. This ensures that we can find a peak element in O(log n) time

```python
def findPeakElement(nums):
    left  = 0
    right = len(nums) - 1
    
    print(nums[left:right+1])

    while left < right:
        mid = (left + right) // 2
        
        if nums[mid] > nums[mid + 1]:
            right = mid
        else:
            left = mid + 1

        print(nums[left:right+1])

    return left
```
