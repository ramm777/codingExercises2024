```python
# Lookup operation in sets is O(1) vs. O(n) in lists, so looups are faster in sets. Lookup ([x in set_a])

def intersection(nums1, nums2):
   
   nums1 = set(nums1)
   nums2 = set(nums2)

   if len(nums1) < len(nums2): 
      return [x for x in nums1 if x in nums2]
   else: 
      return [x for x in nums2 if x in nums1]
   
nums1 = [4,9,5]
nums2 = [9,4,9,8,4]

intersection(nums1, nums2)   
```
