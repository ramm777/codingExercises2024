- Notice, Consequitive means consequitivly increasing [1,2,3, ...]
- Using queue, more intuitive
- O(n) - as each element is processed only once

```python
from collections import deque
```

```python
# Intuitive approach using queue, if you output only the max length of consequitive. See another function. 

def longestConsecutive(nums):
    if not nums:
        return 0

    nums = set(nums)
    max_length = 0

    for num in nums:
        # If (num - 1) is not in the set nums, it means there is no preceding number to num. Therefore, num could be the start of a new sequence.
        if (num - 1) not in nums:  
            queue = deque([num])
            current_length = 0

            while queue:
                current_num = queue.popleft()
                current_length += 1
                
                next_num = current_num + 1
                if next_num in nums:
                    queue.append(next_num)
            
            max_length = max(max_length, current_length)

    return max_length
```

```python
nums1 = [100, 4, 200, 1, 3, 2]
print(longestConsecutive(nums1))  # Output: 4

nums2 = [0, 3, 7, 2, 5, 8, 4, 6, 0, 1]
print(longestConsecutive(nums2))  # Output: 9
```

```python
def longestConsecutive(nums):
    if not nums:
        return 0

    nums = set(nums)
    max_length = 0
    all_subarrays = {}

    for num in nums:
        # If (num - 1) is not in the set nums, it means there is no preceding number to num. Therefore, num could be the start of a new sequence.
        if (num - 1) not in nums:  
            queue = deque([num])
            current_length = 0
            subarray = []

            while queue:
                current_num = queue.popleft()
                current_length += 1
                subarray.append(current_num)
                
                next_num = current_num + 1
                if next_num in nums:
                    queue.append(next_num)
            
            max_length = max(max_length, current_length)
            all_subarrays[current_length] = subarray

    return max_length, all_subarrays[max_length]
```

```python
nums1 = [100, 4, 200, 1, 3, 2, 101,102,103,104]
print(longestConsecutive(nums1)) 
```
