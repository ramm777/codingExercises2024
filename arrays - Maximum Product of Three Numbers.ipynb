- Works by logical reasoning of a sorted array
- Take max of two cases case 1 vs case 2
- Case 1 is logical - a product of biggest values in soted array
- Case 2 is that multiplying two negative numbers results in a positive product. If these two negative numbers are large in magnitude, their product may be greater than the product of the top three positive numbers.

```python
nums = [-1, -2, -3, 4] # (4 * -1 * -2) gives the maximum product = 24


def maximumProduct(nums): 
    nums.sort()

    return max(nums[-1] * nums[-2] * nums[-3], 
               nums[0] * nums[1] * nums[-1]   )
     
maximumProduct(nums)
```

```python
# Instead of sort(), you can use heap to get the n largest or k smallest items

import heapq

nums = [-1, -2, -3, 4] 

print(heapq.nlargest(3, nums))
print(heapq.nsmallest(2, nums))
```
