```python
nums = [1,3,-1,-3,5,3,6,7]

def quicksort(array):

    if len(array) <= 1: 
        return array

    pivot = array[len(array) // 2]

    left = [x for x in array if x < pivot] 
    midd = [x for x in array if x == pivot]
    right= [x for x in array if x > pivot]

    return quicksort(left) + midd + quicksort(right)

quicksort(nums)
```

```python
nums = [1,3,-1,-3,5,3,6,7]
# till what is range()? [3,6,7] last, ind=5 (8-3) or len(nums)-k, will need to add +1 because range() is upuntill 5 so 4

k = 3

def s_w_m(nums, k): 

    m = []
    for i in range(0, len(nums)-k+1):
        
        window = nums[i:i+k]
        window = quicksort(window)
        m.append(window[-1])

    return m       

s_w_m(nums, k)
```
