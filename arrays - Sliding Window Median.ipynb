```python
def quicksort(arr): 
    # Very elegant way of sorting arrays

    if len(arr) <= 1: # important for recursion
        return arr 

    pivot = arr[ len(arr) // 2 ]

    left  = [x for x in arr if x < pivot]  # must not contain '='
    mid   = [x for x in arr if x ==pivot]  # must be exacly like this
    right = [x for x in arr if x > pivot]  # must not contain '='

    return quicksort(left) + mid + quicksort(right)

arr_nums = [3, 6, 8, 10, 1, 2, 1] 
quicksort(arr_nums)
```

```python
def medianSlidingWindow(nums, k):
    
    medians = []
    for i in range(0, len(nums) - k + 1):  # +1 as range() is up until, exclusive
        
        window = nums[i: i+k]
        sorted_window = quicksort(window)
        
        if k % 2 == 0:
            mid_start = (len(sorted_window) // 2) - 1
            mid_end   = (len(sorted_window) // 2)
            median    = (sorted_window[mid_start] + sorted_window[mid_end]) / 2.0
            medians.append(median)
        else:
            median_index = len(sorted_window) // 2
            medians.append(sorted_window[median_index])
            
    return medians
```
