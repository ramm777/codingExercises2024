Move all negative elements at the end of array

- Function traverses through each item, if the item is positive, it swaps it with the current (that can be positive or negative)
- Thus sending it to earlier, as positive indices are slower that current indices

```python
def move_negatives_to_end(nums):
    
    positive = 0  # slower

    for current in range(len(nums)): # faster

        if nums[current] >= 0:  # Check if the current number is positive
            
            nums[positive], nums[current] = nums[current], nums[positive] # if it's positive, swap with current
            print('current:', current, 'positive:', positive, ' |', nums) 
            
            positive += 1

    return nums

nums = [1, 2, -3, -2, 40, 5, -7, -6, 8]
print(move_negatives_to_end(nums))  # Output: [1, 40, 5, 8, -3, -2, -7, -6]
```
