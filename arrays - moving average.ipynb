```python
# Calculate moving average

numbers = [10, 20, 30, 40, 50]
#          0   1   2   3   4
window_size = 3

# end of start idx = 2 [30], after which you must have 2 more == (len(numbers) - window_size)

def movingAverage(numbers, window_size): 

    if window_size > len(numbers) or window_size < 0:
        return('Invalid')


    for i in range(0, len(numbers) - window_size + 1): # add +1 as range() is up until, exclusive 
        print(numbers[i : i + window_size])

        mov_av = sum(numbers[i : i + window_size]) / window_size
        print(mov_av)

movingAverage(numbers, window_size)
```
