```python
import heapq # by default is min heap 

def kClosest(points, k):

    max_heap = []
    for x, y in points:

        distance = -(x**2 + y**2)  # Use negative to convert Min Heap to Max Heap
        
        heapq.heappush(max_heap, (distance, x, y)) # will autmatically create max heap (converted min heap)
        
        # If the heap size exceeds k, remove the point with the largest distance (root). Like in the queue
        if len(max_heap) > k:
            heapq.heappop(max_heap) 
    
    result = [[x, y] for (distance, x, y) in max_heap]
    return result

points2 = [[3, 3], [5, -1], [-2, 4], [0.1, 0.1]]
k2 = 2
print(kClosest(points2, k2))  
```

- Pushing to min heap (with neg dist) will create min heap automatically as a tree. 
- Your function will iteratively delete the root (top) element as it's the largest distance in abs terms and minimum in min heap.
- The heap will automatically reorganised
