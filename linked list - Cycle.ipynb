```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

head = ListNode(3)
head.next = ListNode(2)
head.next.next = ListNode(0)
head.next.next.next = ListNode(-4)
```

```python
current = head
while current is not None: 
    print(current.val)
    current = current.next
```

```python
# Tortoise and Hare Algorithm

def hasCycle(head):

    if not head or not head.next:
        return False

    slow = head
    fast = head.next

    # If the slow and fast meet, it means there's a cycle - return True
    while slow != fast:
        # If the fast reaches the end, it means there's no cycle
        if not fast or not fast.next:
            return False
        
        slow = slow.next
        fast = fast.next.next

    return True
```

- Incorrect to use slow.val != fast.val because two different nodes can have the same value but still be distinct nodes.

- By comparing slow != fast, you are comparing the memory items
