# Sum of Left Leaves
- Don't forget leaves are the end nodes, not the intermediates



```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left   # should start with left
        self.right = right # after left, there is a right
    
# Example tree (not BST):
# root = [3,9,20,null,null,15,7, 10]
#       3
#      / \
#     9  20
#        / \
#       15  7
#       /
#      10  

root = TreeNode(3)      # Initialize the tree with 3, "root" is an isntance of a class
root.left = TreeNode(9)
root.right = TreeNode(20)

# root.left.left = null
# root.left.right = null
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)        
root.right.left.left = TreeNode(10)
```

```python
queue = [root]
sum_lefts = 0

while queue: 

    for _ in range(len(queue)): 
       node = queue.pop(0)

       if node.left is not None: 
            queue.append(node.left)
            if not node.left.left and not node.left.right: # Check if it is a Left leaf 
                sum_lefts += node.left.val
              
       if node.right is not None: 
           queue.append(node.right)    
    print(sum_lefts)       
```

```python
def sumOfLeftLeaves(root): 
    
    if not root: 
        return 0

    queue = [root]
    sum_lefts = 0

    while queue: 

        for _ in range(len(queue)): 
            node = queue.pop(0)

            if node.left is not None: 
                queue.append(node.left)
                
                if not node.left.left and not node.left.right: # Check if it is a Left leaf - sum up 
                    sum_lefts += node.left.val

            if node.right is not None: 
                queue.append(node.right)    
    return sum_lefts 

sumOfLeftLeaves(root)
```

```python
root = TreeNode(3) 
root.left = TreeNode(9)
root.right = TreeNode(20, TreeNode(15), TreeNode(7))

sumOfLeftLeaves(root) # Must be 24
```
