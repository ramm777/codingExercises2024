```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left   # should start with left
        self.right = right # after left, there is a right   
```

```python
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
root.right.left = TreeNode(6)
```

```python
from collections import deque

queue = deque([root])
num_nodes = 1

while queue: 
    node = queue.popleft()

    if node.left: 
        queue.append(node.left)
        num_nodes += 1
    
    if node.right: 
        queue.append(node.right)
        num_nodes += 1
    print(num_nodes)   
```

```python
from collections import deque

def countNodes(root): 

    if not root:
        return 0
    
    queue = deque([root])
    num_nodes = 1

    while queue: 
        node = queue.popleft()

        if node.left: 
            queue.append(node.left)
            num_nodes += 1
        
        if node.right: 
            queue.append(node.right)
            num_nodes += 1
    
    return num_nodes   

countNodes(root)
```
