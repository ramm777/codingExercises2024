```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
# This gives per level sum. Difficult to understand, but look at notes below, and function after this

from collections import deque

def deepestLeavesSum(root):
    if not root:
        return 0
    
    queue = deque([root])
    while queue:
        
        # For each level, reset sum to 0, recalculate level_sum
        level_sum = 0 
        for _ in range(len(queue)):

            node = queue.popleft() # 2 out when [4,5] added, 3 out when [6] added to queue. Next iterate over [4,5,6]
            level_sum += node.val
                        
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        print(level_sum)  
    
    return level_sum
```

```python
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)
root.right.right = TreeNode(6)
root.left.left.left = TreeNode(7)
root.right.right.right = TreeNode(8)

deepestLeavesSum(root)  # Output: 15 (sum of values 7 and 8)
```

```python
from collections import deque

def deepestLeavesSum(root):
    if not root:
        return 0
    
    queue = deque([root])
    levels = []
    while queue:
        
        # For each level, add all possible children to queue. E.g. 2 has [4,5], 3 has [6] 
        level = []
        for _ in range(len(queue)):

            node = queue.popleft() # 2 out when [4,5] added, 3 out when [6] added to queue. Next iterate over [4,5,6] and so on
            level.append(node.val)
                        
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        levels.append(level)      
        print(levels)
    
    return sum(levels[-1])
```
